Q1. Find the Frequency of Each Word in a Sentence

pseudocode:
1. Take a sentence as input.
2. Split the sentence into words.
3. Create an empty dictionary.
4. Count each word.
5. Store each word and its count in the dictionary.
6. Display the dictionary.

Q2. Find the Frequency of Each Character in a Word

pseudocode:
1. Take a word as input.
2. Create an empty dictionary.
3. Read each character.
4. Count how many times each character appears.
5. Store the character and its frequency.
6. Return the dictionary.

Q3. Find the First Non-Repeating Character in a Word

pseudocode:
1. Take a word as input.
2. Count the frequency of each character.
3. Check the characters from left to right.
4. Find the first character whose count is 1.
5. Return that character.
6. If no character exists, return None.


Q4. Check Whether Two Strings Are Anagrams

Pseudocode:
1. Take two strings as input.
2. Sort the characters of both strings.
3. Compare the sorted strings.
4. If both are equal, return True.
5. Otherwise, return False.

Q5. Find the First Repeating Character in a String

Pseudocode:
1. Take a string as input.
2. Create an empty set.
3. Read each character from left to right.
4. Check whether the character is already present.
5. If present, return that character.
6. Otherwise, add the character to the set.
7. If no character repeats, return None.

Q6. Move All Zeros to the End of a List

Pseudocode:
1. Take a list as input.
2. Create an empty list for non-zero numbers.
3. Create another empty list for zeros.
4. Read each number.
5. Separate zeros and non-zero numbers.
6. Join both lists.
7. Return the final list.

Q7. Find the Second Highest Distinct Number in a List

Pseudocode:
1. Take a list as input.
2. Remove duplicate numbers.
3. Sort the numbers in descending order.
4. Check whether at least two distinct numbers exist.
5. Return the second number.
6. Otherwise, return None.

Q8. Remove Duplicates Without Changing Original Order

Pseudocode:
1. Take a list as input.
2. Create an empty result list.
3. Read each element.
4. Check whether the element already exists.
5. If it does not exist, add it to the result.

Q9. Find the Missing Number from 1 to N

Pseudocode:
1. Take a list and N as input.
2. Calculate the expected sum using N*(N+1)/2.
3. Calculate the actual sum of the list.
4. Subtract the actual sum from the expected sum.
5. Return the missing number.

Q10. Find Two Numbers Whose Sum Equals a Target

Pseudocode:
1. Take a list and target as input.
2. Use two loops to check pairs of different positions.
3. Scan the second position from left to right.
4. Check whether the sum equals the target.
5. If it matches, return the pair.
6. If no pair matches, return an empty list.

Q11. Find the Number Closest to Zero

Pseudocode:
1. Take a list of numbers as input.
2. Calculate the absolute value of each number.
3. Find the number with the smallest distance from zero.
4. If two numbers have equal distance, choose the positive number.
5. Return the result.

Q12. Find Common Elements Between Two Lists

Pseudocode:
1. Take two lists as input.
2. Create an empty result list.
3. Read each element from the first list.
4. Check whether it exists in the second list.
5. Avoid adding duplicate elements.
6. Return the result list.

Q13. Rotate a List to the Right by K Positions

Pseudocode:
1. Take a list and K as input.
2. Check whether the list is empty.
3. Calculate K modulo the length of the list.
4. Separate the last K elements.
5. Move them to the beginning.
6. Return the rotated list.

Q14. Find the Maximum Sum of Two Consecutive Elements

Pseudocode:
1. Take a list as input.
2. Calculate the sum of the first two elements.
3. Store it as the maximum sum.
4. Check every consecutive pair.
5. Compare each pair sum with the maximum sum.
6. Update the maximum if a larger sum is found.
7. Return the maximum sum.

Q15. Find All Pairs of Numbers Whose Sum Equals a Target

Pseudocode:

2. Create an empty set to store unique pairs.
3. Use two loops to check different positions.
4. Check whether the sum equals the target.
5. Sort the two numbers in each pair.
6. Store the unique pairs.
7. Sort all pairs by their first element.
8. Return the final list.
